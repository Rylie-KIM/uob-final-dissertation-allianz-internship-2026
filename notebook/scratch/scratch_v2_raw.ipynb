# scratch · v2 raw 확인

**Kernel: `fttl-v2` (env-v2, Python 3.10).** Z:의 v2 pickle을 열어 구조를 확인하는 일회성 노트북.
여기서 확인한 내용은 `01_export_v2.ipynb` / `config.py`에 반영하고, 이 파일 자체는 버려도 됨.

```python
import joblib
import pandas as pd

RAW = r"Z:\...\Data\clean_dataset.pkl"   # 실제 경로 채우기

# .pkl이지만 joblib dump — pd.read_pickle은 "stack_global requires str"로 깨짐 (2026-08-10 확인)
raw = joblib.load(RAW)
print("shape:", raw.shape)
print()
for col in raw.columns:
    print(f"  {col:<40} {raw[col].dtype}")
```

```python
raw.head()
```

```python
# 다른 pkl도 같은 방식으로 열어서 비교 (예: transformed split, production log 등)
OTHER = r"Z:\...\...pkl"   # 경로 채우기

other = joblib.load(OTHER)   # joblib은 순수 pickle도 열 수 있음
print("shape:", other.shape)
print()
for col in other.columns:
    print(f"  {col:<40} {other[col].dtype}")
other.head()
```

```python
# join 실험 — 두 표에서 claim id 컬럼 이름을 확인한 뒤 채우기
ID_RAW, ID_OTHER = "<id in raw>", "<id in other>"

print("raw   id nunique / rows:", raw[ID_RAW].nunique(), "/", len(raw))
print("other id nunique / rows:", other[ID_OTHER].nunique(), "/", len(other))
print("overlap (raw ids in other):", raw[ID_RAW].isin(other[ID_OTHER]).mean().round(4))

joined = raw.merge(other, left_on=ID_RAW, right_on=ID_OTHER, how="inner", suffixes=("", "_other"))
print("joined:", joined.shape)
joined.head()
```
